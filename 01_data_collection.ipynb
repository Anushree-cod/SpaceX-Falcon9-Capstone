# 01 — Data Collection and Wrangling
**Goal:** collect public SpaceX launch data and create a reproducible flat CSV.

This notebook uses the public SpaceX API v4 at run time. API responses can change; inspect missing values before analysis.

```python
from pathlib import Path
import sys
import pandas as pd
ROOT = Path.cwd()
if not (ROOT / "src").exists() and (ROOT.parent / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
from src.spacex_pipeline import collect_launch_table

df = collect_launch_table()
print("Rows, columns:", df.shape)
display(df.head())
```

```python
(ROOT / "data").mkdir(exist_ok=True)
df.to_csv(ROOT / "data" / "spacex_launches_raw.csv", index=False)
df.info()
```

```python
# Data quality checks
quality = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_pct": (100 * df.isna().mean()).round(1),
    "dtype": df.dtypes.astype(str)
}).sort_values("missing_pct", ascending=False)
display(quality.head(20))
print("Duplicate launch names:", df["name"].duplicated().sum())
print("Date range:", df["date_utc"].min(), "to", df["date_utc"].max())
```
